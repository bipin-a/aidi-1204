## Before you start: How to tell distributions apart

Throughout this assignment, every time you describe a scenario you'll answer four questions:

- **What repeats?** (what is one trial/attempt/event?)
- **What stops the process?** (why do you stop?)
- **What are you counting?** (what accumulates?)
- **What stays constant?** (what doesn't change between trials?)

These four questions are a **distribution detector.** The answers tell you exactly which distribution you're dealing with — and more importantly, the *same real-life scenario* can become a *different distribution* just by changing what stops you and what you write down at the end.

Here are three examples, starting simple.

---

### Example 1 — Flipping a coin (simplest case)

You flip a coin 10 times and count how many heads you get.

| Question | Answer |
|---|---|
| **What repeats?** | One coin flip |
| **What stops?** | You've done 10 flips (fixed number) |
| **What accumulates?** | Number of heads |
| **What stays constant?** | 50% chance of heads each flip |

Fixed number of trials, counting successes → **Binomial**

Now change the question: *How many flips until you get the first heads?*

| Question | Answer |
|---|---|
| **What repeats?** | One coin flip |
| **What stops?** | You get heads (first success) |
| **What accumulates?** | Number of flips |
| **What stays constant?** | 50% chance of heads each flip |

Keep going until first success, counting attempts → **Geometric**

Same coin, same flip. Different stopping rule, different distribution.

---

### Example 2 — Free throws (same gym, four distributions)

You're a basketball player who makes 30% of your free throws. Watch how the same scenario creates four different distributions:

**"How many do I make in 20 shots?"**

| Question | Answer |
|---|---|
| **What repeats?** | One free throw |
| **What stops?** | You've taken 20 shots (fixed) |
| **What accumulates?** | Number of makes |
| **What stays constant?** | p = 0.30 |

→ **Binomial** (fixed trials, counting successes)

**"How many shots until my first make?"**

| Question | Answer |
|---|---|
| **What repeats?** | One free throw |
| **What stops?** | You finally make one |
| **What accumulates?** | Number of shots taken |
| **What stays constant?** | p = 0.30 |

→ **Geometric** (keep going until first success, counting attempts)

**"How many shots until I've made 5?"**

| Question | Answer |
|---|---|
| **What repeats?** | One free throw |
| **What stops?** | You've made 5 baskets |
| **What accumulates?** | Number of shots taken |
| **What stays constant?** | p = 0.30 |

→ **Negative Binomial** (keep going until r successes, counting attempts)

Notice: *the geometric is just the negative binomial with r = 1.*

---

### Example 3 — Coffee shop (Poisson and Normal enter the picture)

You manage a coffee shop. On average, 8 customers walk in per hour.

**"How many customers walk in during the next hour?"**

| Question | Answer |
|---|---|
| **What repeats?** | A customer arriving (random event) |
| **What stops?** | The hour ends (fixed time window) |
| **What accumulates?** | Number of arrivals |
| **What stays constant?** | Average rate = 8 per hour |

→ **Poisson** (counting random events in a fixed time window)

Now a different question about the same coffee shop:

**"How much coffee (in mL) does one random cup weigh?"**

Each cup's volume is affected by dozens of tiny factors: how long the barista pours, machine pressure, bean grind size, water temperature, cup positioning, etc. No single factor dominates — they all add up.

| Question | Answer |
|---|---|
| **What repeats?** | Each tiny random factor (pour time, grind, pressure, etc.) |
| **What stops?** | The cup is done — all factors have played their part |
| **What accumulates?** | Their combined effect on the final volume |
| **What stays constant?** | Same machine, same recipe, same process |

→ **Normal** (many small independent factors adding up)

The scenario doesn't pick the distribution. **The question you ask about the scenario picks the distribution.** These four questions force you to be precise about that.

### Cheat sheet for reference

| Distribution | Question it answers | Key parameters |
|---|---|---|
| Binomial | How many successes in `n` fixed trials? | `n` (trials), `p` (probability) |
| Geometric | How many attempts until the first success? | `p` (probability) |
| Poisson | How many events in a fixed time/space window? | `λ` (average rate) |
| Negative Binomial | How many attempts until `r` successes? | `r` (target), `p` (probability) |
| Normal | What value results from many small random effects? | `μ` (mean), `σ` (std dev) |
