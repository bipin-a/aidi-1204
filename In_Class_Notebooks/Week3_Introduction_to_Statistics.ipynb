### 2.4 Strings - Working with Text

**IMPORTANT REMINDER:** Round brackets `()` are used to "call/execute" a method

````python
# String methods
var = 'lower-case-text'
print("Original:", var)

# The .upper() method converts all text to uppercase
print("Uppercase:", var.upper())

# Split converts the string into a list that is split by the argument in the function
print("Split by 'e':", var.split('e'))
````

#### Important: In-place modifications

**In-place** means if the object was modified/overwritten.

In our case, the `.upper()` method does NOT make modifications in place!

````python
var = 'lower-case-text'
var.upper()  # This doesn't change var

print("var is still:", var)

# To keep the change, assign it to a new variable
updatedvar = var.upper()
print("updatedvar is:", updatedvar)
````

#### String Methods - Checking Content

````python
# Check if string contains only digits
print("'2.2'.isdigit():", '2.2'.isdigit())  # False (has a period)
print("'2'.isdigit():", '2'.isdigit())      # True

# Check if numeric (works with unicode numbers)
print("'1.1'.isnumeric():", '1.1'.isnumeric())  # False

# You can use help() or ? to get documentation
# Try: help(str.isdigit)
````

### 2.5 Escape Characters

Special characters in strings:
- `\n`: Newline
- `\t`: Tab
- `\\`: Backslash
- `\'`: Single quote
- `\"`: Double quote

````python
# Examples of escape characters
print("Hello\nWorld")  # Newline
print("Hello\tWorld")  # Tab

# Sometimes we have problems dealing with escape characters
path = "C:\new_folder\file.txt"
print("Problematic path:", path)  # \n is interpreted as newline!
````

One fix to the problem above is to use **raw strings**: `r'str'`

### 2.6 Formatted String Literals (f-strings)

A cool way of embedding another variable inside your string!

Syntax: `f'string_{var_name}'`

Uses a combination of `f` before your string and `{}` wrapped around your variable

### 2.7 Booleans

A binary value of `True` or `False`

A return type when validating a condition

````python
print(type(True))
print(type(False))

# Comparison operations
var = 10
print("var == 100:", var == 100)  # False
print("var != 100:", var != 100)  # True
print("var == 10:", var == 10)    # True
print("var != 9:", var != 9)      # True

# Store result in a variable
mask = var == 100
print("mask:", mask)
````

### 2.8 Boolean Operations: `and`, `or`, `not`

- `and`: checks if both values are valid (not null) or if they are both True
- `or`: checks if any value is valid (not null) or if either is True
- `not`: negates a boolean value

````python
# OR operations
print("True or True:", True or True)      # True
print("False or True:", False or True)    # True
print("False or False:", False or False)  # False

# AND operations
print("\nTrue and True:", True and True)    # True
print("False and True:", False and True)    # False

# NOT operation
print("\nnot False:", not False)  # True
print("not True:", not True)      # False
````

````python
# Truthiness in Python - non-zero/non-empty values are True
if 1 or 10:
    print('At least one value is truthy')

if 10 and [1,1,1]:
    print('Both values are truthy')

if None:
    print('This will not print')  # None is falsy

# Complex expressions
print("\n3 and 9:", 3 and 9)  # Returns 9 (last truthy value)
print("3 and 9 or 7:", 3 and 9 or 7)  # Returns 9
````

### Common pitfalls (booleans and logic)
- `if some_list:` checks if the list is non-empty, not if it contains a value.
- `and`/`or` return one of the operands, not just True or False.
- `==` checks equality; `is` checks identity.

### 2.11 Iterating with Loops

**IMPORTANT NOTE:** Indentation tells Python what should run inside your for-loop

There are 2 ways to iterate:

**Method 1:**
```python
for item in items:
    # do something with item
```

**Method 2:**
```python
for i in range(len(items)):
    item = items[i]
    # do something with item
```

````python
fruits = ['apple', 'banana', 'kiwi']

# Method 1: Direct iteration
print("Method 1:")
for fruit in fruits:
    print(f"  I like {fruit}")

print("\nMethod 2:")
# Method 2: Using index
for i in range(len(fruits)):
    fruit = fruits[i]
    print(f"  Fruit {i}: {fruit}")
````

### Common pitfalls (lists and loops)
- Indexing starts at 0, so the last element is `list[-1]`.
- Modifying a list while iterating over it can skip elements.
- Indentation controls what is inside the loop.

### 2.13 Functions

**Introduction:**
A function is a reusable block of code that performs a specific task. It can take inputs (parameters) and return outputs.

- **Parameters** are variables defined in the function signature
- **Arguments** are the actual values you pass to the function when calling it

````python
# Without a function - repetitive code
name = "Tyson"
print(f"Hello, {name}")

name = "Jack"
print(f"Hello, {name}")

name = "Mike"
print(f"Hello, {name}")
````

````python
# Define a function
def greet(name):
    return f"Hello, {name}!"

# Call/execute the function with ()
print(greet("Alice"))
print(greet("Rob"))
print(greet("Tyson"))
````

#### Return Values

Functions can return values using the `return` statement. The return value can be of any type (string, number, list, etc.) or even `None`.

If no return is specified, Python will automatically return `None`.

````python
def add(a, b):
    print('function is starting')
    total = a + b
    print(total)
    print('function is ending')
    return total

result = add(3, 4)
print("\nResult stored in variable:", result)
````

#### Returning Multiple Values as a Tuple

The most common way to return multiple values is to return them as a tuple. You don't need to explicitly create a tuple using parentheses-Python does this automatically.

````python
def get_coordinates():
    x = 10
    y = 20
    return x, y

# Calling the function
coords = get_coordinates()
print("Coordinates:", coords)
print("Type:", type(coords))
print("X coordinate:", coords[0])
print("Y coordinate:", coords[1])

# Unpacking
outx, outy = get_coordinates()
print(f"\nUnpacked: x={outx}, y={outy}")
````

#### Returning Multiple Values as a List

````python
def get_fruits():
    return ["apple", "banana", "cherry"]

fruits = get_fruits()
print("Fruits:", fruits)
print("Type:", type(fruits))
````

#### Returning Multiple Values as a Dictionary

If you want to return a set of named values (key-value pairs), you can return a dictionary.

````python
def get_person_info():
    return {"name": "Alice", "age": 30, "location": "NYC"}

person_info = get_person_info()
print("Person info:", person_info)
print("Type:", type(person_info))
print("Name:", person_info["name"])

# Using .get() method (safer - returns None if key doesn't exist)
print("Country:", person_info.get('country', 'Canada'))  # Default value
````

### 2.14 Scope

**Scope** refers to the region in the code where a variable is accessible.

**Types of Scope:**
- **Local Scope**: Variables defined within a function
- **Enclosing (Nonlocal) Scope**: Variables defined in outer functions
- **Global Scope**: Variables defined at the top level
- **Built-in Scope**: Python's built-in names like `len`, `int`, etc.

````python
# Example (generally bad practice to use global variables)
global_var = 10

def checking_vars(y):
    print("Global var:", global_var)
    print("Local var y:", y)
    print("ID of global_var:", id(global_var))
    print("ID of y:", id(y))
    return None

checking_vars(global_var)
````

### 2.15 Control Flow

Control flow allows you to make decisions in your code!

#### `if` Statement

Allows you to execute a block of code if a condition is `True` or is not `None`

````python
# Example
x = 10
if x > 5:
    print("x is greater than 5")

# Example with truthiness
x = None
if x:
    print("x is not None")
else:
    print("x is None")
````

#### `else` Statement

Used to execute a block of code when the condition in the `if` statement is not met

````python
x = 3
if x > 5:
    print("x is greater than 5")
else:
    print("x is not greater than 5")
````

#### `elif` Statement

Allows you to check multiple conditions after an `if`

**Note:** The order of the `elif` conditions is important!

````python
x = 15
if x > 20:
    print("x is greater than 20")
elif x > 10:
    print("x is greater than 10 but less than or equal to 20")
elif x > 5:
    print("x is greater than 5 but less than or equal to 10")
else:
    print("x is 5 or less")
````

#### `break` Statement

Used to exit a loop

````python
for i in range(1, 6):
    if i == 3:
        break  # Exits the loop when i equals 3
    print(i)
````

#### `continue` Statement

Used to skip the rest of the current iteration and move to the next one

````python
for i in range(1, 6):
    if i == 3:
        continue  # Skips printing when i equals 3
    print(i)
````

#### `pass` Statement

Does nothing. It's used as a placeholder where you want to leave code for later

````python
x = 10
if x > 5:
    pass  # TODO: Implement this later
else:
    print("x is less than or equal to 5")
````

#### Nested `if` Statements

You can nest `if` statements inside other `if` statements to check more complex conditions

````python
x = 20
if x > 10:
    print("x is greater than 10")
    if x > 15:
        print("x is also greater than 15")
    else:
        print("x is 15 or less")
else:
    print("x is 10 or less")
````

#### Simplifying Nested Conditions

Instead of deeply nested conditions, use logical operators!

````python
# Complex nested version
has_passport = False
has_driving_license = True

if has_passport:
    print("Can travel")
else:
    if has_driving_license:
        print("Can travel")

# Simplified version using OR
if has_passport or has_driving_license:
    print("Can travel (simplified)")
````

````python
# Combining conditions
age = 22
is_student = False
has_special_permission = True

if age > 18 and (is_student or has_special_permission):
    print("Eligible for discount")
````

#### `while` Loop with Control Flow

The `while` loop will continue to execute as long as the condition is `True`

````python
i = 1
while i <= 5:
    print(i)
    i += 1  # Increment i to avoid infinite loop
````

````python
# Using break and continue in a while loop
i = 0
while i < 10:
    i += 1
    if i == 3:
        continue  # Skip the iteration when i equals 3
    if i == 5:
        break  # Exit the loop when i equals 5
    print(i)
````

#### Loop `else` Clause

You can use an `else` clause with loops. The `else` block runs when the loop is NOT terminated by a `break`

````python
# Without break - else executes
for i in range(1, 5):
    print(i)
else:
    print("Loop completed without hitting a break")
````

````python
# With break - else does NOT execute
for i in range(1, 5):
    if i == 3:
        break
    print(i)
else:
    print("This will not be printed because the loop was broken")
````

#### One-line `if` Statements (Ternary Operator)

Python allows you to write compact if-else statements using the ternary operator

Syntax: `<value_if_true> if <condition> else <value_if_false>`

````python
x = 10
result = "x is even" if x % 2 == 0 else "x is odd"
print(result)

# Another example
message = "x is 10" if x == 10 else "x is not 10"
print(message)
````

#### Combining Multiple Conditions

You can combine conditions using `and`/`or`

````python
x = 12
if x > 10 and x < 20:
    print("x is between 10 and 20")

if x == 10 or x == 12:
    print("x is either 10 or 12")
````

### Common pitfalls (control flow)
- Off-by-one errors with `range()` (the stop value is excluded).
- Infinite loops from forgetting to update the loop variable.
- `break` can skip important cleanup if used carelessly.

## Part 5: Explanatory and Response Variables

When studying relationships between variables:
- **Explanatory variable (x)**: The variable that might affect another (also called independent variable)
- **Response variable (y)**: The variable being affected (also called dependent variable)

**Example**: Does age affect substance use?
- Explanatory: age (x)
- Response: substance type (y)

**Important**: Association != Causation! Only randomized experiments can prove causation.

### 5.1 Example: Age vs Days Using Any Substance

````python
# Hypothesis: Does age (explanatory) affect days of substance use (response)?

# Remove missing values
clean_data = drug_data[['age', 'daysanysub']].dropna()

# Scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(clean_data['age'], clean_data['daysanysub'], alpha=0.5)
plt.xlabel('Age (Explanatory Variable)')
plt.ylabel('Days Using Any Substance (Response Variable)')
plt.title('Relationship Between Age and Substance Use Days')
plt.grid(True, alpha=0.3)
plt.show()

# Calculate correlation
correlation = clean_data['age'].corr(clean_data['daysanysub'])
print(f"Correlation coefficient: {correlation:.3f}")
print("\nNote: This shows association, NOT causation!")
````

### 5.2 Example: Housing Status vs Treatment Link

````python
# Does homeless status (explanatory) affect linkage to treatment (response)?

# Create a cross-tabulation
crosstab = pd.crosstab(drug_data['homeless'], drug_data['link'], margins=True)
print("Cross-tabulation: Homeless Status vs Treatment Link")
print(crosstab)

# Visualize
ct_normalized = pd.crosstab(drug_data['homeless'], drug_data['link'], normalize='index') * 100

plt.figure(figsize=(10, 6))
ct_normalized.plot(kind='bar', color=['salmon', 'lightgreen'], edgecolor='black')
plt.xlabel('Housing Status (Explanatory Variable)')
plt.ylabel('Percentage')
plt.title('Treatment Linkage by Housing Status (Response Variable)')
plt.legend(title='Linked to Treatment')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

### Common pitfalls (relationships)
- Correlation does not prove causation.
- Confounders can create misleading relationships.
- Small sample sizes can exaggerate patterns.

## Part 6: Data Collection Principles

### Key Concepts:

1. **Population**: The entire group we want to study
2. **Sample**: A subset of the population we actually collect data from
3. **Anecdotal Evidence**: Evidence from one or two cases (sample size = 1 or 2) - not reliable!
4. **Random Sample**: Every individual has an equal chance of being selected
5. **Biased Sample**: Some individuals are more likely to be selected than others
6. **Non-response Bias**: When many people don't respond to a survey

### 6.1 Simulation: Population vs Sample

````python
# Let's use the Titanic dataset from seaborn as our "population"
titanic = sns.load_dataset('titanic')

print(f"Population size: {len(titanic)} passengers")
print(f"\nPopulation survival rate: {titanic['survived'].mean()*100:.2f}%")

titanic.head()
````

````python
# Take a RANDOM sample
np.random.seed(42)  # For reproducibility
sample_size = 50

random_sample = titanic.sample(n=sample_size, random_state=42)

print(f"Sample size: {len(random_sample)} passengers")
print(f"Sample survival rate: {random_sample['survived'].mean()*100:.2f}%")
print(f"\nDifference from population: {abs(titanic['survived'].mean() - random_sample['survived'].mean())*100:.2f}%")
````

### 6.2 Demonstration: Biased Sample

````python
# What if we only sample first-class passengers? (BIASED!)
biased_sample = titanic[titanic['pclass'] == 1].sample(n=sample_size, random_state=42)

print(f"Biased sample (first class only) size: {len(biased_sample)}")
print(f"Biased sample survival rate: {biased_sample['survived'].mean()*100:.2f}%")
print(f"\nPopulation survival rate: {titanic['survived'].mean()*100:.2f}%")
print(f"\nDifference: {abs(titanic['survived'].mean() - biased_sample['survived'].mean())*100:.2f}%")
print("\nWARNING: This biased sample overestimates survival rate!")
````

````python
# Visualize the bias
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Population
titanic['survived'].value_counts().plot(kind='bar', ax=axes[0], color=['salmon', 'lightgreen'])
axes[0].set_title(f'Population (n={len(titanic)})')
axes[0].set_ylabel('Count')
axes[0].set_xticklabels(['Died', 'Survived'], rotation=0)

# Random sample
random_sample['survived'].value_counts().plot(kind='bar', ax=axes[1], color=['salmon', 'lightgreen'])
axes[1].set_title(f'Random Sample (n={sample_size})')
axes[1].set_xticklabels(['Died', 'Survived'], rotation=0)

# Biased sample
biased_sample['survived'].value_counts().plot(kind='bar', ax=axes[2], color=['salmon', 'lightgreen'])
axes[2].set_title(f'Biased Sample - 1st Class Only (n={sample_size})')
axes[2].set_xticklabels(['Died', 'Survived'], rotation=0)

plt.tight_layout()
plt.show()
````

## Part 7: Sampling Strategies

There are different ways to select a sample from a population:

1. **Simple Random Sampling**: Every individual has equal chance
2. **Stratified Sampling**: Divide into groups (strata), sample proportionally from each
3. **Cluster Sampling**: Divide into clusters, randomly select entire clusters
4. **Multistage Sampling**: Like cluster sampling, but sample within selected clusters

### 7.1 Simple Random Sampling

````python
# Simple random sample - everyone has equal chance
print("SIMPLE RANDOM SAMPLING")
print("=" * 50)

simple_random = drug_data.sample(n=20, random_state=42)
print(f"Sample size: {len(simple_random)}")
print(f"\nSubstance distribution in sample:")
print(simple_random['substance'].value_counts())

print(f"\nSubstance distribution in population:")
print(drug_data['substance'].value_counts())
````

### 7.2 Stratified Sampling

````python
# Stratified sampling - ensure proportional representation from each substance type
print("STRATIFIED SAMPLING")
print("=" * 50)

# Calculate proportion of each substance in population
substance_props = drug_data['substance'].value_counts(normalize=True)
print("Population proportions by substance:")
print(substance_props)

# Stratified sample - sample proportionally from each stratum
sample_size = 40
stratified_sample = drug_data.groupby('substance', group_keys=False).apply(
    lambda x: x.sample(frac=sample_size/len(drug_data), random_state=42)
)

print(f"\nStratified sample size: {len(stratified_sample)}")
print(f"\nSubstance distribution in stratified sample:")
print(stratified_sample['substance'].value_counts())
print(f"\nProportions in stratified sample:")
print(stratified_sample['substance'].value_counts(normalize=True))
````

````python
# Compare simple random vs stratified
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Population
drug_data['substance'].value_counts().plot(kind='bar', ax=axes[0], color='skyblue', edgecolor='black')
axes[0].set_title('Population')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=45)

# Simple random sample
simple_random['substance'].value_counts().plot(kind='bar', ax=axes[1], color='lightcoral', edgecolor='black')
axes[1].set_title('Simple Random Sample')
axes[1].tick_params(axis='x', rotation=45)

# Stratified sample
stratified_sample['substance'].value_counts().plot(kind='bar', ax=axes[2], color='lightgreen', edgecolor='black')
axes[2].set_title('Stratified Sample')
axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

print("Notice: Stratified sampling maintains the proportions better!")
````

### 7.3 Cluster Sampling

````python
# Cluster sampling - divide by race groups, select entire clusters
print("CLUSTER SAMPLING")
print("=" * 50)

print("Available clusters (race groups):")
print(drug_data['racegrp'].value_counts())

# Randomly select 2 clusters (race groups)
np.random.seed(42)
available_clusters = drug_data['racegrp'].unique()
selected_clusters = np.random.choice(available_clusters, size=2, replace=False)

print(f"\nRandomly selected clusters: {selected_clusters}")

# Include ALL members from selected clusters
cluster_sample = drug_data[drug_data['racegrp'].isin(selected_clusters)]

print(f"\nCluster sample size: {len(cluster_sample)}")
print(f"\nRace distribution in cluster sample:")
print(cluster_sample['racegrp'].value_counts())
````

### 7.4 Multistage Sampling

````python
# Multistage sampling - select clusters, then sample within clusters
print("MULTISTAGE SAMPLING")
print("=" * 50)

# Stage 1: Select clusters (race groups)
print(f"Stage 1 - Selected clusters: {selected_clusters}")

# Stage 2: Randomly sample within each selected cluster
multistage_sample = cluster_sample.groupby('racegrp', group_keys=False).apply(
    lambda x: x.sample(n=min(10, len(x)), random_state=42)
)

print(f"\nStage 2 - Sample 10 from each cluster")
print(f"\nMultistage sample size: {len(multistage_sample)}")
print(f"\nRace distribution in multistage sample:")
print(multistage_sample['racegrp'].value_counts())
````

### Common pitfalls (sampling strategies)
- Cluster samples can be cheaper but less representative.
- Stratified samples require correct strata labels.
- Non-random selection inside strata reintroduces bias.

## Part 8: Experimental Design

### Key Principles:

1. **Controlling**: Ensure differences in outcomes are due to treatment, not other factors
2. **Randomization**: Randomly assign subjects to treatment/control groups
3. **Replication**: Collect many observations to strengthen evidence
4. **Blocking**: Account for variables that might affect the response

### Observational Study vs Experiment:
- **Observational Study**: Just observe, no intervention -> Can show association
- **Experiment**: Actively assign treatments -> Can prove causation

### 8.1 Simulation: The Importance of Randomization

````python
# Simulate a drug treatment experiment
print("EXPERIMENT SIMULATION: Testing a New Treatment")
print("=" * 50)

# Create a subset of drug data for our "experiment"
experiment_data = drug_data[['id', 'age', 'sex', 'substance', 'cesd']].dropna().copy()
experiment_data = experiment_data.head(100)  # Use 100 subjects

print(f"Total subjects in experiment: {len(experiment_data)}")
print(f"\nBaseline depression score (CESD):")
print(experiment_data['cesd'].describe())
````

````python
# RANDOMIZATION: Randomly assign to treatment or control
np.random.seed(42)
experiment_data['group'] = np.random.choice(['Treatment', 'Control'], size=len(experiment_data))

print("Group assignment:")
print(experiment_data['group'].value_counts())

# Check if randomization balanced the groups
print("\nAverage age by group (should be similar if randomization worked):")
print(experiment_data.groupby('group')['age'].mean())

print("\nGender distribution by group:")
print(pd.crosstab(experiment_data['group'], experiment_data['sex']))
````

````python
# Simulate treatment effect
# Let's say treatment reduces depression score by 10 points on average
experiment_data['post_cesd'] = experiment_data['cesd'].copy()

# Apply treatment effect with some random variation
treatment_mask = experiment_data['group'] == 'Treatment'
experiment_data.loc[treatment_mask, 'post_cesd'] = (
    experiment_data.loc[treatment_mask, 'post_cesd'] - 10 + np.random.normal(0, 5, treatment_mask.sum())
)

# Control group also has some natural variation
control_mask = experiment_data['group'] == 'Control'
experiment_data.loc[control_mask, 'post_cesd'] = (
    experiment_data.loc[control_mask, 'post_cesd'] + np.random.normal(0, 5, control_mask.sum())
)

# Calculate change
experiment_data['change'] = experiment_data['cesd'] - experiment_data['post_cesd']

print("Results after treatment:")
print(experiment_data.groupby('group')['change'].describe())
````

````python
# Visualize the results
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Before treatment
experiment_data.boxplot(column='cesd', by='group', ax=axes[0])
axes[0].set_title('Depression Scores BEFORE Treatment')
axes[0].set_xlabel('Group')
axes[0].set_ylabel('CESD Score')
plt.sca(axes[0])
plt.xticks([1, 2], ['Control', 'Treatment'])

# After treatment
experiment_data.boxplot(column='post_cesd', by='group', ax=axes[1])
axes[1].set_title('Depression Scores AFTER Treatment')
axes[1].set_xlabel('Group')
axes[1].set_ylabel('CESD Score')
plt.sca(axes[1])
plt.xticks([1, 2], ['Control', 'Treatment'])

plt.suptitle('')  # Remove automatic title
plt.tight_layout()
plt.show()

print("\nOK: The treatment group shows lower depression scores after treatment!")
print("OK: Because we used randomization, we can infer causation.")
````

### 8.2 Blocking Example

````python
# BLOCKING: Account for sex differences
# Maybe males and females respond differently to treatment

print("BLOCKING BY SEX")
print("=" * 50)

# Randomize within each sex (block)
blocked_experiment = drug_data[['id', 'age', 'sex', 'cesd']].dropna().copy()
blocked_experiment = blocked_experiment.head(100)

# Randomize within each block
np.random.seed(42)
blocked_experiment['group'] = blocked_experiment.groupby('sex')['sex'].transform(
    lambda x: np.random.choice(['Treatment', 'Control'], size=len(x))
)

print("Group assignment by sex (block):")
print(pd.crosstab(blocked_experiment['sex'], blocked_experiment['group']))
print("\nThis ensures equal representation in both treatment groups for each sex!")
````

### Common pitfalls (experiments)
- Non-random assignment can bias results.
- Too few participants per group can hide true effects.
- Changing multiple factors at once makes it hard to isolate cause.
